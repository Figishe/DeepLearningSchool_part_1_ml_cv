Перед решением любой задачи важно понимать, как будет оцениваться ваше решение. В данном случае мы используем стандартную для задачи классификации метрику ROC-AUC. Именно поэтому она очень удобна для соревнований.

Посчитать ее легко:

Данные не расходятся, и HasPhoneService = no всегда приводит к значениям no phone service в другой колонке, а значит значения зависимы и будут мешать обучению линейной модели. Поэтому устраняем зависимость через замену.

Также в данных значение **No internet service** в зависимых от инета столбцах может встречаться только если **HasInternetService = No,** проверим что это и правда так:

```python
has_internet = data['HasInternetService'] != 'No'
internet_dependent_columns = data.columns[data.apply(lambda col: (col == 'No internet service').any())].tolist()
no_internet_counts = (data[internet_dependent_columns] == 'No internet service').sum(axis=1)
display(f"Internet-dependent: {internet_dependent_columns}")

if ((no_internet_counts != 0) & (no_internet_counts != len(internet_dependent_columns))).any():
    display("NOT OK, some columns say 'No internet service' while others don't")
else:
    display("OK, all 'No internet service' are in sync between each other")

if (len(data[has_internet & (no_internet_counts > 0)]) > 0):
    display("NOT OK")
else:
    display("OK, 'No internet service' values are in sync with HasInternetService = No")
```

Значения 'No internet service' в таком случае зависимы от HasInternetService, и тоже мешают обучению лин. моделей. Устраняем зависимость через замену:

```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
```

Нормирование числовых признаков

```python
scaler = StandardScaler()
scaler.fit(data[num_cols])

norm = scaler.transform(data[num_cols])
norm_df = pd.DataFrame(norm)
norm_df.columns = [x + "_scaled" for x in num_cols]
```

Кодирование категориальных признаков

```python
# label_encoder = LabelEncoder()
# label_encoder.fit(data[cat_cols])

one_hot_encoder = OneHotEncoder(drop='first')
one_hot_encoder.fit(data[cat_cols])

cat_encoded_df = pd.DataFrame(
    one_hot_encoder.transform(data[cat_cols]).toarray(),
    columns=one_hot_encoder.get_feature_names_out(cat_cols)
)
```

Объединение закодированных данных

```python
data_transformed = pd.concat([norm_df, cat_encoded_df], axis=1)

if target_col in data.columns:
    data_transformed = pd.concat([data_transformed, data[target_col].reset_index()], axis=1)

if 'index' in data_transformed.columns:
    data_transformed = data_transformed.drop(columns=['index'])
```

```python
X = data_transformed.drop(target_col, axis=1)
y = data_transformed[target_col]

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

model_lr = LogisticRegressionCV(
    scoring='roc_auc', 
    refit=True, 
    solver='saga', 
    penalty='elasticnet', 
    l1_ratios=[0, 0.00001, 0.0001, 0.001, 0.01, 0.1], 
    Cs=[0.001, 0.01, 0.1, 1, 10, 100], 
    cv=7,
    max_iter=7500, 
    n_jobs=-1,
    random_state=42
)
model_lr.fit(X_train, y_train)
```

```python
print("- Score:", model_lr.score(X_test, y_test))
print("- Best C:", model_lr.C_[0])
print("- Best L1 ratio:", model_lr.l1_ratio_[0])
print("- max_iter:", model_lr.max_iter)
```
