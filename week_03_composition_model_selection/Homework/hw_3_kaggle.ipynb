```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
```

Нормирование числовых признаков

```python
if train:
    scaler = StandardScaler()
    scaler.fit(data[num_cols])
else:
    # Test данные не используются для scaler fit, иначе утечка
    pass
```

```python
norm = scaler.transform(data[num_cols])
norm_df = pd.DataFrame(norm)
norm_df.columns = [x + "_scaled" for x in num_cols]
```

Кодирование категориальных признаков

```python
one_hot_encoder = OneHotEncoder(drop='first')
one_hot_encoder.fit(data[cat_cols])

cat_encoded_df = pd.DataFrame(
    one_hot_encoder.transform(data[cat_cols]).toarray(),
    columns=one_hot_encoder.get_feature_names_out(cat_cols)
)
```

Объединение закодированных данных

```python
data_transformed = pd.concat([norm_df, cat_encoded_df], axis=1)

if target_col in data.columns:
    data_transformed = pd.concat([data_transformed, data[target_col].reset_index()], axis=1)

if 'index' in data_transformed.columns:
    # Где-то выше индекс превратился в колонку, чтобы долго не разбираться - дроп
    data_transformed = data_transformed.drop(columns=['index'])
```

```python
if train:
    X = data_transformed.drop(target_col, axis=1)
    y = data_transformed[target_col]

    X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

    model_lr = LogisticRegressionCV(
        scoring='roc_auc', 
        refit=True, 
        solver='saga', 
        penalty='elasticnet', 
        l1_ratios=[0, 0.002, 0.005, 0.007, 0.01, 0.1], 
        Cs=[0.01, 0.1, 1, 10, 100], 
        cv=10,
        max_iter=7500, 
        n_jobs=-1,
        random_state=43
    )
    model_lr.fit(X_train, y_train)

    print("- Score:", model_lr.score(X_test, y_test))
    print("- Best C:", model_lr.C_[0])
    print("- Best L1 ratio:", model_lr.l1_ratio_[0])
    print("- max_iter:", model_lr.max_iter)

else:
    pass # готовая модель просто применяется ниже
```

```python
from catboost import CatBoostClassifier
from sklearn.model_selection import PredefinedSplit, StratifiedKFold

if train:
    X = data.drop(target_col, axis=1)
    y = data[target_col]

    X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=4, stratify=y)

    cat_model = CatBoostClassifier(
        eval_metric='AUC',
        iterations=600,
        early_stopping_rounds=30,
        thread_count=1,
        verbose=False,
    )

    param_grid = {
        'learning_rate': [0.005, 0.007, 0.01, 0.03, 0.05],
        'depth': [2, 3, 4],
        'l2_leaf_reg': [0, 0.01, 0.03, 0.05, 0.07, 0.1],
        'random_strength': [1, 5, 10],
    }

    useCV = True

    if not useCV:
        # CV это кратно дороже, тут эксперименты с одним ручным split

        # В GridSearchCV уходят train + test вместе, поэтому их надо склеить
        X_full = pd.concat([X_train.reset_index(drop=True), X_test.reset_index(drop=True)], ignore_index=True)
        y_full = pd.concat([y_train.reset_index(drop=True), y_test.reset_index(drop=True)], ignore_index=True)

        # test fold при этом один константный, чтобы не раздувать обучение
        test_fold = [-1]*len(X_train) + [0]*len(X_test)
        ps = PredefinedSplit(test_fold)

        grid = GridSearchCV(cat_model, param_grid, scoring='roc_auc', cv=ps, n_jobs=16, refit=True, verbose=2)
        grid.fit(X_full, y_full, cat_features=cat_cols)
    else:
        # Если не жалко времени
        cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
        grid = GridSearchCV(cat_model, param_grid, scoring='roc_auc', cv=cv, n_jobs=16, refit=True, verbose=2)
        grid.fit(X, y, cat_features=cat_cols)

    print("best params:", grid.best_params_)
    print("best CV roc_auc:", grid.best_score_)

    best = grid.best_params_
    best_cat_model = CatBoostClassifier(
        iterations=5000,
        learning_rate=best['learning_rate'],
        depth=best['depth'],
        l2_leaf_reg=best['l2_leaf_reg'],
        random_strength=best['random_strength'],
        eval_metric='AUC',
        early_stopping_rounds=200,
        use_best_model=True,
        random_seed=42,
        verbose=100,
        thread_count=-1
    )

    best_cat_model.fit(
        X_train, y_train,
        eval_set=(X_test, y_test),
        cat_features=cat_cols
    )

    print("Final AUC:", best_cat_model.get_best_score()['validation']['AUC'])
```
